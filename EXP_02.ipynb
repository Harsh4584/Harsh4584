```python
b=0
w=0
lr = 0.01
```

```python
y = df['package']
x = df['cgpa']
```

```python
df.count()
```

```python
m = len(df)

for i in range(200):
    y_pred = b + w * x
    cost = ((y - y_pred) ** 2).sum() / m

    dw = (-2 / m) * ((y - y_pred) * x).sum()
    db = (-2 / m) * (y - y_pred).sum()


    w = w - lr * dw
    b = b - lr * db
    print(cost)
```
